# Maya's helper, with memory switches

This is Maya's Mugs help desk from Episode 2 of Bit By Bot, with three new switches. You don't type any code here.

1. Press the play button on the cell below.
2. Type a message in the box and press Enter. To finish the chat, press Enter on an empty box.
3. Tick a switch, then press play again. Every press of play starts a new, empty chat.

Under each reply you'll see two boxes:

- **SENT TO AI** is what the program sent the AI this turn. The briefing and shipping policy are shortened to one line here; the AI gets them in full.
- **SAVED CONVERSATION** is what the program has written down for the next turn.

Replies come from Colab's free built-in AI, so yours may be worded differently from the video.


```python
#@title Maya's helper with memory switches
#@markdown Tick a switch, then press play. Every press of play starts a new, empty chat.
save_customer_messages = False #@param {type:"boolean"}
send_saved_conversation = False #@param {type:"boolean"}
save_helper_replies = False #@param {type:"boolean"}

from google.colab import ai

briefing = """
You are the help desk for Maya's Mugs, a small online mug shop.
Be short and friendly. Write plain sentences, no lists.
Answer only from the shipping policy below.
A country inside a listed region counts as that region.
If the answer is not in the policy, say you don't know.
"""

policy = """
Shipping policy:
United States: 2 to 4 days
Canada: 5 to 8 days
Europe: 7 to 12 days
Australia: 10 to 14 days
"""

def show(title, lines):
  print("  " + title)
  for line in lines:
    print("  | " + line)

saved = ""
for turn in range(10):
  question = input("Customer: ")
  if question == "":
    break
  prompt = briefing + policy
  if send_saved_conversation:
    prompt = prompt + saved
  prompt = prompt + "Customer: " + question
  answer = ai.generate_text(prompt).strip()
  answer = answer.removeprefix("Helper:").strip()
  print("Helper:", answer)
  if save_customer_messages:
    saved = saved + "Customer: " + question + "\n"
  if save_helper_replies:
    saved = saved + "Helper: " + answer + "\n"
  print()
  sent = prompt[len(briefing + policy):].splitlines()
  show("SENT TO AI: " + str(len(prompt)) + " characters", ["(briefing and shipping policy)"] + sent)
  show("SAVED CONVERSATION", saved.splitlines() or ["(nothing saved)"])
  print()
print("Chat ended. Press play to start a new chat.")
```
